### Estadísticas UNA 2020-2024

El objetivo principal de este mini proyecto es automatizar la descarga de los datos estadisticos de UNA, convertirlos de PDF a CSV para un mejor analisis 

#### Web Scrapping
Ingresamos a la sgte página donde encotraremos enlaces a los PDFs de Postulantes, Ingresantes, Matriculados, Egresados con los datos de nuestro interes "https://www.una.py/la-universidad/estadisticas/poblacion-carreras-de-grado-por-sedes"

#### Obtener los links automaticamente

```python
import requests
from bs4 import BeautifulSoup


url = "https://www.una.py/la-universidad/estadisticas/poblacion-carreras-de-grado-por-sedes"

response = requests.get(url) # obtener datos del servidor una.py
soup = BeautifulSoup(response.text, "html.parser") #convierte html a objeto

pdf_links = []

for link in soup.find_all('a'):
    href = link.get('href')

    if href and '.pdf' in href:
        pdf_links.append(href)

pdf_links
```

### Descargamos los PDFs obtenidos

```python
from pathlib import Path
import requests

# Carpeta destino
pdf_folder = Path('Documents/Estadistica-UNA')
pdf_folder.mkdir(parents=True, exist_ok=True)       # crea carpeta directorio

for pdf_url in pdf_links:
    file_name = pdf_url.split('/')[-1]

    response = requests.get(pdf_url)

    with open(pdf_folder/file_name, "wb") as f:      # descarga y guarda el archivo, escribiendo directo en disco
        f.write(response.content)

    print(f'Descargado: {file_name}')
```

#### Extraemos las tablas de los PDFs
Importamos tabula para leer los PDFs y convertirlas a CSV para posterior análisis

```python
import tabula

pdf_path = 'Documents/Estadistica-UNA/Postulantes-por-Unidad-Academica-2024'

tables = tabula.read_pdf(pdf_path, pages='all', multiple_tables=True)

len(tables)
```

#### Guardamos cada tabla
Utilizamos pandas para trabajar con dataframes

```python
import pandas as pd

for i, table in enumerate(tables):
    table.to_csv(f'Documents/Estadistica-UNA/csv/tabla_{i}.csv',index=False)
    
```

#### Unimos las tablas

```python
df = pd.concat(tables, ignore_index=True)
df
```
